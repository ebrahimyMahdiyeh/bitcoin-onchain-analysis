### BTC price 
The approximately linear trend of the log‑transformed price indicates that the underlying growth of Bitcoin was exponential, which is why using the logarithmic scale makes the dynamics easier and more reliable to analyze.
The slope of the log‑price changes significantly around three to four key points, two of which align with the halving cycles in 2012 and 2016. These halving events effectively shifted the slow downward drift in price into strong upward momentum, and in the following months the market experienced substantial rallies.
There are also two major early‑period crashes — mid‑2011 and early 2014 — each driven by different structural shocks:
Mid‑2011 crash:
Caused primarily by the Mt. Gox security breach (June 2011), where a hacked account caused a massive flash crash down to $0.01, exposing severe exchange vulnerabilities and triggering a loss of market confidence.
Early 2014 crash:
Driven by the collapse of Mt. Gox, at the time the largest Bitcoin exchange, which halted withdrawals and later declared bankruptcy after losing roughly 850,000 BTC. This event led to a prolonged and deep bear market due to systemic fear and liquidity loss.
By adding moving averages to the price chart, the long‑term structure of Bitcoin’s market cycles becomes clearer, since short‑term volatility is smoothed out and the underlying trend can be observed more reliably. The behavior of the moving averages shows that the slope of the long‑term price trend often weakens or flattens before halving events, while after a halving the slope tends to increase again, indicating the beginning of a new growth regime. During major crashes or bearish periods, slow moving averages also capture the gradual loss of market momentum more clearly than the raw price, which tends to fluctuate sharply. In these situations, the flattening or reduction in the slope of the moving average reflects a structural weakening of the trend. In addition, points where the price crosses the slow moving average often signal major market phase transitions: a crossover from below to above typically indicates the beginning of a sustained bullish phase, while a crossover from above to below usually reflects growing weakness and the transition into a bearish phase.

### Hash rate
Unlike the price, the hash rate shows a predominantly upward and steady trend without sharp peaks, extreme spikes, or deep valleys. This stability reflects the continuously increasing security of the Bitcoin network and its resilience to short‑term market shocks.
Over the long term, the slope of hash rate growth appears to be indirectly dependent on Bitcoin’s price. When Bitcoin becomes more valuable, mining becomes more profitable, attracting additional miners and increasing the network’s total computational power. As a result, periods of sustained price growth are often accompanied by stronger expansion in hash rate.
By examining the moving averages of the hash rate, two additional patterns become clearer. The use of moving averages is particularly helpful because it smooths out short‑term noise and volatility, making it easier to detect genuine structural changes in the underlying trend. In particular, the slope of the moving averages tends to temporarily decrease and flatten both around halving events and during bearish market phases. Around halvings, the sudden reduction in block rewards lowers miner revenue, which can reduce mining profitability and force less efficient miners to shut down their equipment. In earlier stages of the network, such as the first halving, the raw hash rate even experienced a brief decline. In later periods, however, the hash rate typically continues to rise while the moving averages reveal a slowdown in its growth. A similar effect occurs during bearish market phases, when falling Bitcoin prices compress miner profitability and slow the expansion of mining activity. In these situations, the hash rate usually maintains its long‑term upward trend, but the slope of its growth becomes weaker. As profitability improves and market conditions recover, the growth rate of the hash rate typically accelerates again.

### Unique Active Adress 
Active addresses show a long‑term upward expansion that moves closely with Bitcoin’s price but with a much smoother profile. During bull markets, the metric forms brief parabolic bursts, reflecting rapid spikes in network activity and user engagement. In bear markets, it declines more mildly than the price, indicating that the baseline network usage persists even when speculation fades.
Because shifts in this metric often appear slightly earlier than the corresponding moves in price, active addresses can act as a light but reliable early‑indicator of both local tops and bottoms.

### Median Confirmation Time
Median Confirmation Time, unlike the previous three metrics, exhibits no long‑term trend and is inherently trend‑free. It is neither exponential nor cyclical; instead, it is a purely spike‑driven congestion indicator. The metric only reacts when the network becomes saturated. During bullish periods, rising transaction demand and a congested mempool produce sharp, short‑lived spikes. In bearish periods, as transactional pressure decreases, the metric quickly returns to low‑volatility baseline levels.

🗣️ Because of its reactive nature and its exclusive dependence on network congestion, Median Confirmation Time has no structural relationship with Hash Rate or Difficulty. Its sole purpose is to quantify the immediate load placed on Bitcoin’s limited block‑space capacity.
