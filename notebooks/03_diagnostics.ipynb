```python
import os
import sys
import joblib

# --------------------------
# Detect project root
# --------------------------
def find_project_root(start_path):
    current = start_path
    while True:
        if os.path.exists(os.path.join(current, "src")) and os.path.exists(os.path.join(current, "results")):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            raise RuntimeError("Project root not found")
        current = parent

project_root = find_project_root(os.getcwd())
print("Detected project root:", project_root)

sys.path.append(os.path.join(project_root, "src"))

# --------------------------
# Load fitted VECM model
# --------------------------
model_path = os.path.join(project_root, "results", "vecm_model.pkl")

print("Loading model from:", model_path)
vecm_res = joblib.load(model_path)

print("Model loaded successfully!")
```

The final results show that the **VECM** performs well in terms of stability and identifying a long-run relationship, but the residuals continue to display problematic behavior. The **Portmanteau test** rejects the null hypothesis of no residual autocorrelation, and this issue persists even after exploring different lag lengths, including lag 8. This suggests that the issue likely originates from the structure of the data rather than the lag specification.

The **Jarque–Bera normality test** is also strongly rejected. The residual distribution—especially for the first variable—shows very high kurtosis and positive skewness, which is typical for financial data, especially cryptocurrencies, and does not necessarily invalidate the long-run component of the model.

On the positive side, all **characteristic roots** lie inside the unit circle, confirming that the model is dynamically stable. The **error-correction coefficients (alpha)** are statistically significant, indicating that both variables adjust to deviations from the long-run equilibrium. The difference in adjustment speeds is also reasonable and economically meaningful.

Since modifying the lag length did not resolve the residual issues, further analysis should focus on the **data structure**. There is a strong possibility of **structural breaks** or **multiple market regimes**. Splitting the data into separate phases—such as bull, bear, and recovery markets—often improves residual behavior and model accuracy. If the goal is volatility analysis or forecasting, a combined approach such as **VECM-GARCH** may be appropriate.

Overall, the current model is suitable for long-run analysis, but the limitations related to residual behavior should be acknowledged when interpreting the results.
